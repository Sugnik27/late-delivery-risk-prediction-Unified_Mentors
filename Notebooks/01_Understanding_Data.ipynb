##  Preliminary Feature Importance Analysis

### Why do this before modeling?
Before building any model, it is useful to understand which features have the strongest statistical relationship with the target variable. This guides our cleaning and feature engineering decisions.

- **Numerical features** → Pearson Correlation with target
- **Categorical features** → Cramér's V (measures association between categorical variable and binary target)

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
numerical_cols = [col for col in numerical_cols if col != 'Late_delivery_risk']

correlations = df[numerical_cols].corrwith(df['Late_delivery_risk']).abs().sort_values(ascending=False)

corr_table = pd.DataFrame({
    'Feature': correlations.index,
    'Absolute Correlation with Target': correlations.values.round(4)
})

print("Numerical Feature Correlation with Late_delivery_risk (sorted):")
display(corr_table)
```

```python
# Cramér's V Function and Categorical Importance

def cramers_v(col, target):
    confusion_matrix = pd.crosstab(col, target)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = len(col)
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

cramers_scores = {}
for col in categorical_cols:
    try:
        score = cramers_v(df[col], df['Late_delivery_risk'])
        cramers_scores[col] = round(score, 4)
    except:
        cramers_scores[col] = None

cramers_table = pd.DataFrame({
    'Feature': cramers_scores.keys(),
    "Cramér's V with Target": cramers_scores.values()
}).dropna().sort_values("Cramér's V with Target", ascending=False).reset_index(drop=True)

print("Categorical Feature Association with Late_delivery_risk (Cramér's V):")
display(cramers_table)
```
